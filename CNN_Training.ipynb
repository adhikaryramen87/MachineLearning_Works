```python
import tensorflow as tf
from tensorflow.keras import layers, models

def build_keras_standard_block():
  model = models.Sequential([
      # 1. convolution: 32 filters, 3*3 kernels
      layers.Conv2D(32, (3, 3), padding = 'same', input_shape = (28, 28, 1)),
      # 2. Batch Normalization: Normalize the 32 feature maps
      layers.BatchNormalization(),
      # 3. Activation Function: ReLU after normalization
      layers.Activation('relu'),
      # 4. Pooling: Reduces 28*28 to 14*14
      layers.MaxPooling2D(pool_size=(2,2)),

      # convert the 2D to 1D
      layers.Flatten(),
      layers.Dense(10, activation = 'softmax')
  ]
  )

  return model
```

```python
model = build_keras_standard_block()
model.summary()
```

MLP Approach

```python
mlp_model = models.Sequential([
    layers.Flatten(input_shape=(28, 28, 1)),
    layers.Dense(512, activation = 'relu'),
    layers.Dense(10, activation='softmax')
])

mlp_model.summary()
```

CNN Approach

```python
cnn_model = models.Sequential([
    layers.Conv2D(32, kernel_size=(3,3), activation='relu', input_shape = (28, 28, 1)),
    layers.Flatten(),
    layers.Dense(10, activation='softmax')
])

cnn_model.summary()
```
