Import all the required libraries

```python
import pandas as pd
import numpy as np
```

Load the data

```python
data = pd.read_csv('/content/spam.csv')
data.head()
```

```python
data.Category.value_counts()
```

```python
data['spam'] = data['Category'].apply(lambda x:1 if x == 'spam' else 0)
data.head()
```

Data Split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(data.Message, data.spam, stratify=data.spam, test_size=0.2, random_state=42)
```

```python
X_train[:4]
```

```python
X_train.shape
```

Message -> Feature Vectorization

```python
X_train.values
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer()
X_train_cv = vectorizer.fit_transform(X_train.values)
type(X_train_cv)
```

```python
X_train_cv
```

4457 => number of messages in the dataframe

7663 => number of unique words/features in the messages

```python
X_train_cv.toarray()
```

```python
vectorizer.get_feature_names_out()[3000:6000]
```

```python
vectorizer.vocabulary_
```

```python
vectorizer.get_feature_names_out()[1743]
```

```python
y_train
```

```python
from sklearn.naive_bayes import MultinomialNB
model = MultinomialNB()
model.fit(X_train_cv, y_train)
```

```python
X_test_cv = vectorizer.transform(X_test.values)
y_pred = model.predict(X_test_cv)
y_pred
```

FP => Prediction = Positive Actual = Negative

Means the model predicts the mail is spam but actually the mail is important and urgent to respond.

```python
from sklearn.metrics import precision_score
precision = precision_score(y_test, y_pred)
print(precision)
```

```python
from sklearn.pipeline import Pipeline
clf = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('nb', MultinomialNB())
])
```

```python
clf.fit(X_train, y_train)
```

```python
emails = [
    'Hi Priya, can we have a meeting tomorrow?',
    'Upto 20% OFF on H&M. Dont miss this exclusive reward'
]

emails_vectorizer = vectorizer.transform(emails)
model.predict(emails_vectorizer)
```

Raw TF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

reviews = [
    'The plot is good.',
    'The plot is good, good, really good.'
]

tf_vectorizer = TfidfVectorizer(use_idf=False)

tf_matrix = tf_vectorizer.fit_transform(reviews)
tf_matrix.toarray()
df_tf_vectorizer = pd.DataFrame(tf_matrix.toarray(), columns = tf_vectorizer.get_feature_names_out())

df_tf_vectorizer
```

'is', 'the', 'a' [high frequency] => tf_vectorizer values will be also high

```python
from sklearn.feature_extraction.text import TfidfVectorizer

reviews = [
    'The plot is good.',
    'The plot is good, good, really good.'
    'The plot is a masterpiece',
    'I hate the plot',
    'The plot is okay'
]

tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(reviews)
tfidf_matrix.toarray()
df_tfidf_vectorizer = pd.DataFrame(tfidf_matrix.toarray(), columns = tfidf_vectorizer.get_feature_names_out())

df_tfidf_vectorizer
```
