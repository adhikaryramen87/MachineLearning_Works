Import all the required libraries

```python
import gensim
import nltk
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize

nltk.download('punkt_tab')
```

Preprocessed the text corpus

```python
text = [
    "NLP is transforming the AI industry",
    "Machine Learning, Deep Learning and NLP are powerful domains in Data Science",
    "AI is changing the future",
    "Deep Learning based neural networks helps in NLP specific tasks"
]

processed_text = [word_tokenize(sentence.lower()) for sentence in text]

print(processed_text)
```

Train the word2vec -> from scratch based on your input data [processed_text]

```python
# default implementation or training is done via CBOW
# you want to train via skip gram, sg = 1
model = Word2Vec(
    sentences = processed_text,
    vector_size = 100,
    window = 5,
    min_count = 1,
    workers = 4
)
```

Word vector for a specific word

```python
print(model.wv['nlp'])
```

Similar words wrt nlp

```python
print(model.wv.most_similar('nlp'))
```

Similarity score between two words

```python
print(model.wv.similarity('ai', 'nlp'))
```
